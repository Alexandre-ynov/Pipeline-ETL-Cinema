```python
import pandas as pd
```

```python
csv_file_path = "../data/raw/movies.csv"
json_file_path = "../data/raw/movies_API.json"

def load_csv(csv_file_path):
    try:
        df_csv = pd.read_csv(csv_file_path, encoding='utf8')
        print(f"CSV download successful : {csv_file_path}")
        return df_csv
    except FileNotFoundError:
        print(f"Error : File not found -> {csv_file_path}")
    except Exception as e:
        print(f"Error : {e}")

def load_json(json_file_path):
    try:
        df_json= pd.read_json(json_file_path, encoding='utf8')
        print(f"Json download successful : {json_file_path}")
        return df_json
    except FileNotFoundError:
        print(f"Error : File not found -> {json_file_path}")
    except Exception as e:
        print(f"Error: {e}")
```

```python
if __name__ == "__main__" :
    df_csv = load_csv(csv_file_path)
    df_json = load_json(json_file_path)
```
