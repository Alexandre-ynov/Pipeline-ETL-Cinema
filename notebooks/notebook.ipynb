```python
import pandas as pd
```

```python
csv_file_path = "../data/raw/movies.csv"
json_file_path = "../data/raw/movies_API.json"

def load_csv(csv_file_path):
    try:
        df_csv = pd.read_csv(csv_file_path, encoding="utf-8")
        print(f"CSV download successful: {csv_file_path}")
        return df_csv
    except FileNotFoundError:
        print(f"Error: file not found -> {csv_file_path}")
    except Exception as error:
        print(f"Error loading CSV: {error}")

def load_json(json_file_path):
    try:
        api_data = pd.read_json(json_file_path, encoding="utf-8")
        df_json = pd.json_normalize(api_data["results"])
        print(f"JSON download successful: {json_file_path}")
        return df_json
    except FileNotFoundError:
        print(f"Error: file not found -> {json_file_path}")
    except Exception as error:
        print(f"Error loading JSON: {error}")
```
